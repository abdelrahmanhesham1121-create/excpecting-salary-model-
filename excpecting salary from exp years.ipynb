```python
x = data[['Experience Years']]
y = data[['Salary']]
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model.fit(x,y)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
model.rank_
```

```python
model.predict([[7]])
```

```python
plt.scatter(x,y)
plt.plot(x,model.predict(x),color='r')
```

```python
model.score(x,y)
```
